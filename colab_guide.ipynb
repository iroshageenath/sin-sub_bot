```python
import os, glob, zipfile, shutil
zip_path = '/content/uploads/anime_sinhala_bot_v9.zip'
extract_root = '/content/anime_bot_extract'
shutil.rmtree(extract_root, ignore_errors=True)
os.makedirs(extract_root, exist_ok=True)
with zipfile.ZipFile(zip_path) as z: z.extractall(extract_root)
project_dir = next(root for root, dirs, files in os.walk(extract_root) if 'bot.py' in files and 'requirements.txt' in files)
print('PROJECT:', project_dir)
```

## Colab Secrets
Add `BOT_TOKEN`, `ADMIN_USER_ID`, and `GEMINI_API_KEY_1`. Optionally add `_2` and `_3`.

```python
from google.colab import userdata
import os
for name in ['BOT_TOKEN','ADMIN_USER_ID'] + [f'GEMINI_API_KEY_{i}' for i in range(1, 11)]:
    try:
        value = userdata.get(name)
        if value: os.environ[name] = value
    except Exception: pass
print('BOT_TOKEN:', bool(os.environ.get('BOT_TOKEN')))
print('ADMIN_USER_ID:', bool(os.environ.get('ADMIN_USER_ID')))
print('GEMINI_API_KEY_1:', bool(os.environ.get('GEMINI_API_KEY_1')))
```
